```python
import time
from concurrent.futures import ProcessPoolExecutor, ThreadPoolExecutor
from multiprocessing import Pool, cpu_count

cpu_count = cpu_count()
cpu_count
```

```python
def multithreading_concurrent(func, args):
    with ThreadPoolExecutor() as pool:
        res = pool.map(func, args)
    return list(res)

def multiprocessing_concurrent(func, args):
    with ProcessPoolExecutor() as pool:
        res = pool.map(func, args)
    return list(res)

def multiprocessing_native(func, args):
    with Pool() as pool:
        res = pool.map(func,args)
    return list(res)
```

```python
marker = time.time()
load_url(urls[0])
print("Call :", time.time() - marker)
```

```python
marker = time.time()
multithreading_concurrent(load_url, urls)
print("Multithreading :", time.time() - marker)
```

```python
marker = time.time()
multiprocessing_concurrent(load_url, urls)
print("Multiprocessing :", time.time() - marker)
```

```python
marker = time.time()
multiprocessing_native(load_url, urls)
print("multiprocessing_native :", time.time() - marker)
```

```python
from itertools import repeat
marker = time.time()
multithreading_concurrent(load_url, zip(urls,repeat(0),repeat(0),repeat(0)))
print("multiprocessing_native :", time.time() - marker)
```
